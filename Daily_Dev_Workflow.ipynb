# Daily Development Workflow (`NeuralForceField`)

This notebook is a quick reminder for daily Git operations in this repo.

- Repo path: `/data2/users/chuanyul/NeuralForceField`
- Working branch: `chuanyul-branch-forked_NeuralForceField`
- Personal remote: `personal` (public GitHub)
- Upstream remote: `upstream` (original MIT repo)
- Group remote: `origin` (lab repo)


## 1. Start of Day

```bash
cd /data2/users/chuanyul/NeuralForceField
git switch chuanyul-branch-forked_NeuralForceField
git branch --show-current
git status
git remote -v
git fetch upstream
```


## 2. Commit Local Progress

```bash
git add -A
git commit -m "WIP: <short message>"
```


## 3. Push to Personal Public Repo

```bash
git push personal chuanyul-branch-forked_NeuralForceField
```


## 4. Pull Useful Upstream Updates

Pick a target upstream branch (example shown with `surface-sampling-0.3.0`):

```bash
git fetch upstream
git switch chuanyul-branch-forked_NeuralForceField
git merge upstream/surface-sampling-0.3.0
# resolve conflicts if needed
git push personal chuanyul-branch-forked_NeuralForceField
```


## 5. Optional Safety

```bash
git config remote.pushDefault personal
```

Then plain `git push` goes to your personal repo by default.


## 6. Safety Rules

- Always check branch before editing: `git branch --show-current`
- Do daily work on your branch, not directly on `master`
- Merge upstream updates when needed, not blindly
